```python
calculated_revenue = (
    df['quantity'] * 
    df['unit_price'] * 
    (1 - df['discount']))
df['revenue_difference'] = df['revenue'] - calculated_revenue
df['revenue_difference'].describe()
```

```python
total_customers = df['customer_id'].nunique()
total_customers
```

data transformation

```python
df['year'] = df['order_date'].dt.year
df['year'].value_counts().sort_index()
```

```python
df['month'] = df['order_date'].dt.month
df['month'].unique()
```

```python
df['month_name'] = df['order_date'].dt.month_name()
df['month_name'].value_counts()
```

```python
df['discount_amount'] = (
    df['quantity'] * 
    df['unit_price'] * 
    df['discount']
)
```

```python
df['gross_sales'] = (
    df['quantity'] * 
    df['unit_price']
)
```
